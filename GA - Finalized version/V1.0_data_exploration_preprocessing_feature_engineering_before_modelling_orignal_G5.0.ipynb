```python
# Training data: October 2022 – September 2025
train_df = df[
    (df['time'] >= '2022-10-01') &
    (df['time'] < '2025-10-01')
].copy()

# Testing data: October 2025 – September 2026
test_df = df[
    (df['time'] >= '2025-10-01') &
    (df['time'] < '2026-10-01')
].copy()
```

```python
train_df.tail()
```

```python
test_df.tail()
```

```python
selected_features = [
    "time",
    "temperature_2m",
    "dew_point_2m",
    "cloud_cover",
    "weather_code",
    "precipitation",
    "hour",
    "month",
    "us_aqi"
]
```

```python
df_weather_train = train_df[selected_features]
df_weather_test = test_df[selected_features]
```

```python
df_weather_train.columns
```

```python
import numpy as np

# Apply cyclic encoding to training data
train_df['hour_sin'] = np.sin(2 * np.pi * train_df['hour'] / 24)
train_df['hour_cos'] = np.cos(2 * np.pi * train_df['hour'] / 24)

train_df['month_sin'] = np.sin(2 * np.pi * train_df['month'] / 12)
train_df['month_cos'] = np.cos(2 * np.pi * train_df['month'] / 12)


# Apply cyclic encoding to testing data
test_df['hour_sin'] = np.sin(2 * np.pi * test_df['hour'] / 24)
test_df['hour_cos'] = np.cos(2 * np.pi * test_df['hour'] / 24)

test_df['month_sin'] = np.sin(2 * np.pi * test_df['month'] / 12)
test_df['month_cos'] = np.cos(2 * np.pi * test_df['month'] / 12)
```

```python
print("Training Data:")
print(train_df[['hour', 'month', 'hour_sin', 'hour_cos',
                'month_sin', 'month_cos']].head())

print("\nTesting Data:")
print(test_df[['hour', 'month', 'hour_sin', 'hour_cos',
               'month_sin', 'month_cos']].head())
```

```python
df_weather_train.to_csv("kathmandu_weather_train_orignal.csv", index = False)
df_weather_test.to_csv("kathmandu_weather_test_orignal.csv", index = False)
```
